# When Does Nighttime Phone Use Become a Sleep Problem?

I started this project with a simple question: does spending more time on my phone before bed go along with worse sleep? As I explored the data, I realized that the amount of time might be only part of the story. The type of app, other habits, and combinations of behaviors might matter too.

1. **Exposure:** How does increasing bedtime screen time relate to sleep?
2. **Context:** Do brightness, app choice, caffeine, and blue-light filters change that relationship?
3. **Stacking:** What happens when several potentially unfavorable habits occur together?
4. **Sleep:** How do REM, deep sleep, and sleep debt differ across those profiles?
5. **Tomorrow:** How do those sleep differences correspond to next-day fatigue?

This notebook brings together the analyses in `story.ipynb` and adds a sleep-stage comparison across the same behavioral profiles.

### Getting ready and understanding the data

I use the saved file `data/bedtime_screentime_sleep_debt.csv`, originally obtained from the Kaggle dataset `samartalwar/sleep-debt-and-screen-time-late-night-phone-habits`. I have not independently verified how the records or sleep-debt categories were created. I treat the results as patterns in this dataset, rather than evidence about everyone’s sleep.

Run this notebook from top to bottom, from the project folder. It uses pandas, NumPy, Matplotlib, seaborn, statsmodels, Patsy, and IPython. The saved CSV means that no new download is needed.

A few terms helped me read the results:

- **Unadjusted** means I compare the recorded values directly. **Adjusted** means a model also accounts for the listed characteristics, such as age and caffeine use. Adjustment cannot account for things the dataset did not measure.
- A **95% confidence interval** shows uncertainty around an estimate under the model’s assumptions. If a difference’s interval includes zero, I do not describe it as a clear difference.
- **REM/deep percentages** describe the share of sleep in a stage. **REM/deep minutes** also depend on how long someone slept. I calculate those minutes from hours and percentages; they are not a separate measurement.
- **Sleep debt** is a recorded category here, not a known number of hours owed.

The models compare different participants. They do not show what would happen if I changed my own habits. I use robust (HC3) standard errors so the uncertainty calculation allows the spread of the errors to vary across observations.

### Looking beyond REM percentage

I split bedtime phone use into ranges to make the first comparisons easier to read. These cutoffs organize the chart; they are not proven safe or unsafe limits. I also keep screen time as a continuous measure in the models below.

I compare total sleep, stage percentages, stage minutes, and all four debt categories. The tables show group sizes and averages. The boxplots show the spread of values, with outliers hidden for readability. The stacked bars show percentages of participants in each debt category.

```python
# Create the complete-case frame used for the phone, sleep, and fatigue comparisons.
phone_story = df.copy()
story_required = ['user_id', 'bedtime_phone_minutes', 'total_sleep_hours', 'rem_sleep_pct',
    'deep_sleep_pct', 'sleep_debt_category', 'next_day_fatigue_score', 'primary_bedtime_app',
    'screen_brightness_pct', 'blue_light_filter_active', 'caffeine_post_5pm_mg',
    'physical_activity_min', 'age', 'gender', 'occupation_type', 'chronotype']
phone_story = phone_story.dropna(subset=story_required).copy()
assert phone_story.user_id.is_unique, 'Repeated users require clustered inference.'
phone_story['screen_30'] = phone_story.bedtime_phone_minutes / 30
phone_story['rem_minutes'] = phone_story.total_sleep_hours * 60 * phone_story.rem_sleep_pct / 100
phone_story['deep_minutes'] = phone_story.total_sleep_hours * 60 * phone_story.deep_sleep_pct / 100
phone_story['severe_debt'] = phone_story.sleep_debt_category.eq('Severe Sleep Debt').astype(int)
story_debt_order = ['Optimal Recovery', 'Mild Deficit', 'Moderate Debt', 'Severe Sleep Debt']
assert set(phone_story.sleep_debt_category) <= set(story_debt_order)
story_ranges = ['0–30', '>30–60', '>60–90', '>90–120', '>120']
phone_story['screen_range'] = pd.cut(phone_story.bedtime_phone_minutes,
    [0, 30, 60, 90, 120, np.inf], labels=story_ranges, include_lowest=True)
story_outcomes = {'total_sleep_hours': 'Total sleep (hours)', 'rem_sleep_pct': 'REM (%)',
    'deep_sleep_pct': 'Deep sleep (%)', 'rem_minutes': 'REM (minutes)', 'deep_minutes': 'Deep sleep (minutes)'}
story_controls = ('C(primary_bedtime_app) + screen_brightness_pct + blue_light_filter_active + '
    'caffeine_post_5pm_mg + physical_activity_min + age + C(gender) + C(occupation_type) + C(chronotype)')
story_rhs = 'screen_30 + ' + story_controls
story_notes = {}
print(f'{len(phone_story):,} rows retained; {len(df) - len(phone_story)} excluded for missing fields.')
story_baseline = phone_story[list(story_outcomes)].agg(['mean', 'median', 'std']).T
story_screen_summary = phone_story.groupby('screen_range', observed=True)[list(story_outcomes)].mean()
story_screen_summary.insert(0, 'n', phone_story.groupby('screen_range', observed=True).size())
story_debt_shares = pd.crosstab(phone_story.screen_range, phone_story.sleep_debt_category,
    normalize='index').reindex(columns=story_debt_order) * 100
display(story_baseline.round(2), story_screen_summary.round(2), story_debt_shares.round(1))

fig, axes = plt.subplots(2, 3, figsize=(16, 9))
for ax, (outcome, label) in zip(axes.flat, story_outcomes.items()):
    sns.boxplot(data=phone_story, x='screen_range', y=outcome, color='#83b6d5', showfliers=False, ax=ax)
    ax.set(xlabel='Bedtime phone use (minutes)', ylabel=label)
    ax.tick_params(axis='x', rotation=25)
story_debt_shares.plot.bar(stacked=True, ax=axes.flat[-1], colormap='YlOrRd')
axes.flat[-1].set(xlabel='Bedtime phone use (minutes)', ylabel='Participants (%)', title='Recorded sleep-debt categories')
axes.flat[-1].legend(fontsize=8)
fig.suptitle('How do sleep outcomes differ across phone-use durations?')
plt.tight_layout()
plt.show()
```

### Comparing another 30 minutes of phone use

Next I fit a model for each sleep measure. I compare the simple screen-time relationship with a model that also includes app category, brightness, filtering, caffeine, activity, age, gender, occupation, and chronotype. Chronotype describes a person’s morning/evening tendency.

For severe sleep debt I use a model for a yes/no outcome. Its odds ratio is not a probability ratio and does not tell me how many extra hours of sleep debt someone has. These models assume a straight-line relationship for the continuous outcomes, and a straight-line relationship with log-odds for severe debt.

```python
story_sleep_models, story_effect_rows = {}, []
for outcome, label in story_outcomes.items():
    for adjustment, rhs in [('Unadjusted', 'screen_30'), ('Adjusted', story_rhs)]:
        model = smf.ols(f'{outcome} ~ {rhs}', data=phone_story).fit(cov_type='HC3')
        if adjustment == 'Adjusted':
            story_sleep_models[outcome] = model
        low, high = model.conf_int().loc['screen_30']
        story_effect_rows.append({'outcome': label, 'model': adjustment,
            'difference per +30 phone minutes': model.params['screen_30'],
            '95% CI lower': low, '95% CI upper': high})
story_screen_effects = pd.DataFrame(story_effect_rows).set_index(['outcome', 'model'])
display(story_screen_effects.round(3))

# For debt, model the probability of the recorded severe category rather than ordinal codes.
story_debt_model = smf.glm('severe_debt ~ ' + story_rhs, data=phone_story,
    family=sm.families.Binomial()).fit(cov_type='HC3')
story_sleep_models['severe_debt'] = story_debt_model
debt_or = np.exp(story_debt_model.params['screen_30'])
debt_or_ci = np.exp(story_debt_model.conf_int().loc['screen_30'])
display(pd.DataFrame({'odds ratio per +30 phone minutes': [debt_or],
    '95% CI lower': [debt_or_ci.iloc[0]], '95% CI upper': [debt_or_ci.iloc[1]]}))

def story_effect_text(outcome):
    row = story_screen_effects.loc[(story_outcomes[outcome], 'Adjusted')]
    return (f"{row.iloc[0]:+.2f} (95% CI {row.iloc[1]:+.2f} to {row.iloc[2]:+.2f})")

story_notes['part1'] = (
    '**What this shows:** For each additional 30 minutes of bedtime phone use, adjusted '
    'total sleep differs by **' + story_effect_text('total_sleep_hours') + ' hours**, '
    'REM duration by **' + story_effect_text('rem_minutes') + ' minutes**, and deep-sleep '
    'duration by **' + story_effect_text('deep_minutes') + ' minutes**. The REM-percentage '
    'difference is ' + story_effect_text('rem_sleep_pct') + ' percentage points; the '
    'deep-sleep-percentage difference is ' + story_effect_text('deep_sleep_pct') + ' percentage points. '
    'Thus duration and sleep composition can tell different stories. Derived stage minutes '
    'are calculated from total sleep and stage percentage, not an independent measurement. '
    f'The adjusted odds ratio for the severe-debt category is {debt_or:.2f} '
    f'(95% CI {debt_or_ci.iloc[0]:.2f}–{debt_or_ci.iloc[1]:.2f}) per 30 minutes. '
    'An odds ratio is not a probability ratio. These models assume linear continuous-outcome '
    'slopes and linear log-odds for severe debt; intervals use HC3 robust standard errors. '
    'A category association does not reveal how many hours of debt accumulate.')
```

```python
display(Markdown(
    '**What I notice:** For another 30 minutes of bedtime phone use, the adjusted total-sleep difference is **'
    + story_effect_text('total_sleep_hours') + ' hours**. The REM-minute difference is **'
    + story_effect_text('rem_minutes') + '** and the deep-sleep-minute difference is **'
    + story_effect_text('deep_minutes') + '**. The numbers in parentheses are 95% confidence intervals.\n\n'
    + 'The REM-percentage difference is ' + story_effect_text('rem_sleep_pct')
    + ' percentage points. This helps me see why a similar percentage does not necessarily mean a similar amount of time asleep. '
    + f'The severe-debt odds ratio is {debt_or:.2f} (95% CI {debt_or_ci.iloc[0]:.2f}–{debt_or_ci.iloc[1]:.2f}). '
    + 'I see associations here, but I cannot identify an exact minute when phone use becomes harmful.'))
```

## 2. Context: Do other habits change the relationship?

After looking at duration, I want to know whether the way someone uses their phone matters. I separate two questions that I initially found easy to mix up:

1. At the **same 60 minutes** of phone use, do the predicted sleep outcomes differ across settings or apps?
2. Does the estimated change in REM percentage for **another 30 minutes** differ across those conditions?

The first compares sleep levels. The second compares slopes, which is what an **interaction** tests. Two groups can have different average sleep levels without having different slopes.

### Comparing conditions at the same duration

For each comparison, I set phone use to 60 minutes and change one factor. I leave each participant’s other recorded characteristics as they were, make predictions, and average them. This gives each condition the same background mix of participants.

I use News / Reading as the app reference, 25% versus 75% brightness, filter OFF versus ON, and 0 versus 100 mg of evening caffeine. These are comparison choices, not recommendations. Each model includes the factor’s interaction with screen time as well as the original adjustment variables.

The counts near one hour help me see how much data are available there. All eligible participants, not just those near one hour, are used to fit the models. Rare combinations can still make a prediction depend on assumptions.

```python
# Return an average adjusted prediction and its gradient for uncertainty/paired contrasts.
def story_standardize(model, settings, probability=False):
    scenario = phone_story.assign(screen_30=2, **settings)
    # statsmodels 0.15+ uses model_spec; earlier versions use design_info.
    design_info = getattr(model.model.data, 'model_spec', None)
    if design_info is None:
        design_info = model.model.data.design_info
    design = np.asarray(build_design_matrices([design_info], scenario)[0])
    if probability:
        prediction = np.asarray(model.predict(scenario))
        value = prediction.mean() * 100
        gradient = (design * (prediction * (1 - prediction))[:, None]).mean(axis=0) * 100
    else:
        gradient = design.mean(axis=0)
        value = float(gradient @ model.params)
    return value, gradient

def story_interval(model, value, gradient):
    se = np.sqrt(max(float(gradient @ np.asarray(model.cov_params()) @ gradient), 0))
    return value - 1.96 * se, value + 1.96 * se

story_conditions = {
    'app': ('screen_30:C(primary_bedtime_app)', 'News / Reading',
        {name: {'primary_bedtime_app': name} for name in sorted(phone_story.primary_bedtime_app.unique())}),
    'brightness': ('screen_30:screen_brightness_pct', '25% brightness',
        {'25% brightness': {'screen_brightness_pct': 25}, '75% brightness': {'screen_brightness_pct': 75}}),
    'filter': ('screen_30:blue_light_filter_active', 'Filter OFF',
        {'Filter OFF': {'blue_light_filter_active': 0}, 'Filter ON': {'blue_light_filter_active': 1}})
}
story_condition_models, story_predictions, story_contrasts = {}, {}, {}

def story_compare_conditions(factor):
    interaction, reference_label, scenarios = story_conditions[factor]
    predictions, contrasts = [], []
    for outcome in list(story_outcomes) + ['severe_debt']:
        probability = outcome == 'severe_debt'
        label = 'Severe debt probability (%)' if probability else story_outcomes[outcome]
        formula = f'{outcome} ~ {story_rhs} + {interaction}'
        model = (smf.glm(formula, data=phone_story, family=sm.families.Binomial()).fit(cov_type='HC3')
                 if probability else smf.ols(formula, data=phone_story).fit(cov_type='HC3'))
        story_condition_models[(factor, outcome)] = model
        ref_value, ref_gradient = story_standardize(model, scenarios[reference_label], probability)
        for condition, settings in scenarios.items():
            value, gradient = story_standardize(model, settings, probability)
            low, high = story_interval(model, value, gradient)
            predictions.append({'condition': condition, 'outcome': label, 'adjusted mean at 60 min': value,
                '95% CI lower': low, '95% CI upper': high})
            if condition != reference_label:
                difference = value - ref_value
                low, high = story_interval(model, difference, gradient - ref_gradient)
                contrasts.append({'condition': condition, 'reference': reference_label, 'outcome': label,
                    'difference at 60 min': difference, '95% CI lower': low, '95% CI upper': high})
    story_predictions[factor] = pd.DataFrame(predictions)
    story_contrasts[factor] = pd.DataFrame(contrasts)
    display(story_predictions[factor].pivot(index='condition', columns='outcome', values='adjusted mean at 60 min').round(2))
    display(story_contrasts[factor].set_index(['condition', 'outcome']).round(3))
    fig, axes = plt.subplots(1, 3, figsize=(16, 5))
    for ax, label in zip(axes, ['Total sleep (hours)', 'REM (minutes)', 'Deep sleep (minutes)']):
        plot = story_predictions[factor].query('outcome == @label').reset_index(drop=True)
        ax.errorbar(plot['adjusted mean at 60 min'], np.arange(len(plot)),
            xerr=[plot['adjusted mean at 60 min'] - plot['95% CI lower'], plot['95% CI upper'] - plot['adjusted mean at 60 min']],
            fmt='o', capsize=3)
        ax.set(yticks=np.arange(len(plot)), yticklabels=plot.condition, xlabel=label)
    fig.suptitle(f'At 60 phone minutes: adjusted sleep outcomes by {factor} (95% intervals)')
    plt.tight_layout()
    plt.show()

def story_condition_note(factor):
    rows = story_contrasts[factor]
    sentences = []
    for _, row in rows.loc[rows.outcome == 'REM (minutes)'].iterrows():
        sentences.append(f"**{row['condition']} vs {row['reference']}**: {row['difference at 60 min']:+.2f} "
            f"REM minutes (95% CI {row['95% CI lower']:+.2f} to {row['95% CI upper']:+.2f}).")
    return ('**At the same 60-minute duration:** ' + ' '.join(sentences) +
        ' The tables also show total sleep, deep sleep, stage percentages, and severe-debt probability. '
        'For the probability outcome, contrasts are percentage points. Intervals that include zero '
        'do not clearly distinguish the conditions. These are model-based associations, not the '
        'expected causal effect of switching a setting or app. Intervals are pointwise and are not '
        'corrected for the many outcome/condition comparisons; app rankings are exploratory.')
```

```python
# App/category: compare like-duration use rather than each app's raw average.
story_near_hour = phone_story.loc[phone_story.bedtime_phone_minutes.between(45, 75)]
display(story_near_hour.groupby('primary_bedtime_app').size().rename('observations at 45–75 phone minutes').to_frame())
story_compare_conditions('app')
story_notes['app'] = story_condition_note('app')
```

```python
# Brightness: keep screen time at one hour and compare 25% with 75% brightness.
display(story_near_hour.screen_brightness_pct.describe().to_frame('brightness near one hour'))
story_compare_conditions('brightness')
story_notes['brightness'] = story_condition_note('brightness')
```

```python
# Blue-light filter: standardize ON/OFF predictions to the same covariate distribution.
display(story_near_hour.groupby('blue_light_filter_active').size().rename('observations at 45–75 phone minutes').to_frame())
story_compare_conditions('filter')
story_notes['filter'] = story_condition_note('filter')
```

```python
# Add caffeine to the same-duration comparison as well as the slope comparison below.
story_conditions['caffeine'] = (
    'screen_30:caffeine_post_5pm_mg', '0 mg after 5 pm',
    {'0 mg after 5 pm': {'caffeine_post_5pm_mg': 0},
     '100 mg after 5 pm': {'caffeine_post_5pm_mg': 100}})
display(story_near_hour.caffeine_post_5pm_mg.describe().to_frame('caffeine near one hour'))
story_compare_conditions('caffeine')

for factor in ['app', 'brightness', 'filter', 'caffeine']:
    rows = story_contrasts[factor]
    sentences = []
    for _, row in rows.loc[rows.outcome == 'REM (minutes)'].iterrows():
        sentences.append(
            f"{row['condition']} compared with {row['reference']}: **{row['difference at 60 min']:+.2f} REM minutes** "
            f"(95% CI {row['95% CI lower']:+.2f} to {row['95% CI upper']:+.2f}).")
    display(Markdown('**My same-duration comparison — ' + factor + ':** ' + ' '.join(sentences)))
```

I read these as differences between model predictions, not the benefit I would get from changing a setting. The severe-debt contrasts use **percentage points**. The intervals apply to each comparison separately; they have not been corrected for the many outcomes and conditions. I therefore treat apparent app rankings as exploratory.

The dataset does not record clock bedtime, when phone use started or stopped, or the gap between putting the phone down and falling asleep. I cannot use these data to test those timing questions.

### Does the REM-percentage slope change?

I now return to the second question. I use REM percentage as the outcome, matching the interaction analysis in the original notebook. This is a narrower question than overall sleep quality.

The baseline model gives one screen-time slope. The next four models let that slope vary with brightness, filter status, caffeine, or app. The line charts hold the other characteristics at the same typical profile. Shaded bands show uncertainty in the fitted average, not the range of individual sleep outcomes.

```python
app = df['primary_bedtime_app']

brightness = df['screen_brightness_pct']
blue_light_filter = df['blue_light_filter_active']
caffeine = df['caffeine_post_5pm_mg']

# Reuse the Series already created above; use one complete-case sample for every model.
rem_analysis = df.assign(
    rem_pct=rem, screen_30=bed_minutes / 30,
    brightness_25=(brightness - 50) / 25,
    filter_on=blue_light_filter, caffeine_100=caffeine / 100,
    bedtime_app=app)
rem_columns = ['rem_pct', 'screen_30', 'brightness_25', 'filter_on', 'caffeine_100',
               'bedtime_app', 'physical_activity_min', 'age', 'gender', 'occupation_type', 'chronotype']
rem_analysis = rem_analysis[rem_columns].dropna().copy()
assert df.loc[rem_analysis.index, 'user_id'].is_unique, 'Repeated users require clustered inference.'
print(f'Analyzing {len(rem_analysis):,} of {len(df):,} rows; {len(df) - len(rem_analysis)} excluded for missing values.')
print('Filter values:', sorted(rem_analysis.filter_on.unique()))
assert set(rem_analysis.filter_on.unique()) <= {0, 1}
rem_main_formula = ('rem_pct ~ screen_30 + brightness_25 + filter_on + caffeine_100 + '
                    'C(bedtime_app) + physical_activity_min + age + C(gender) + '
                    'C(occupation_type) + C(chronotype)')
rem_models, rem_interpretations, rem_interaction_tests = {}, {}, {}

# A fixed typical profile makes the line plots comparable; only the displayed exposures vary.
rem_profile = rem_analysis.iloc[[0]].copy()
for column in rem_columns:
    rem_profile[column] = (rem_analysis[column].median() if pd.api.types.is_numeric_dtype(rem_analysis[column])
                           else rem_analysis[column].mode().iloc[0])
rem_profile['filter_on'] = 0

def rem_design(model, frame):
    # statsmodels 0.15+ uses model_spec; earlier versions use design_info.
    design_info = getattr(model.model.data, 'model_spec', None)
    if design_info is None:
        design_info = model.model.data.design_info
    return np.asarray(build_design_matrices([design_info], frame)[0])

def rem_slope_vector(model, settings):
    low = rem_profile.assign(screen_30=2, **settings)
    high = rem_profile.assign(screen_30=3, **settings)
    return (rem_design(model, high) - rem_design(model, low))[0]

def rem_contrast(model, vector):
    test = model.t_test(vector)
    low, high = np.asarray(test.conf_int()).ravel()
    return {'REM pp per +30 min': float(np.asarray(test.effect).item()),
            '95% CI lower': float(low), '95% CI upper': float(high),
            'p-value': float(np.asarray(test.pvalue).item())}

def rem_slope_table(model, scenarios):
    return pd.DataFrame({label: rem_contrast(model, rem_slope_vector(model, settings))
                         for label, settings in scenarios.items()}).T

def rem_interaction_test(model, key):
    terms = [i for i, name in enumerate(model.params.index) if ':' in name]
    restriction = np.eye(len(model.params))[terms]
    result = model.wald_test(restriction, use_f=True, scalar=True)
    p = float(result.pvalue)
    rem_interaction_tests[key] = {'interaction p-value': p, 'adjusted R2': model.rsquared_adj}
    return p

def rem_lines(model, scenarios, title):
    fig, ax = plt.subplots(figsize=(10, 5))
    for label, settings in scenarios.items():
        grid = pd.concat([rem_profile] * 80, ignore_index=True)
        grid['screen_30'] = np.linspace(rem_analysis.screen_30.min(), rem_analysis.screen_30.max(), 80)
        for column, value in settings.items():
            grid[column] = value
        prediction = model.get_prediction(grid).summary_frame()
        x = grid.screen_30.to_numpy() * 30
        line, = ax.plot(x, prediction['mean'], label=label)
        ax.fill_between(x, prediction['mean_ci_lower'], prediction['mean_ci_upper'], alpha=0.10, color=line.get_color())
    ax.set(xlabel='Bedtime phone use (minutes)', ylabel='Predicted REM sleep (%)', title=title)
    ax.legend(title='Other covariates fixed at the same typical profile')
    plt.tight_layout()
    plt.show()

def rem_effect_sentence(result):
    return (f"{result['REM pp per +30 min']:+.3f} REM percentage points "
            f"(95% CI {result['95% CI lower']:+.3f} to {result['95% CI upper']:+.3f})")
```

```python
rem_unadjusted = smf.ols('rem_pct ~ screen_30', data=rem_analysis).fit(cov_type='HC3')
rem_baseline = smf.ols(rem_main_formula, data=rem_analysis).fit(cov_type='HC3')
rem_models['baseline'] = rem_baseline
rem_baseline_results = pd.DataFrame({
    'Unadjusted': rem_contrast(rem_unadjusted, rem_slope_vector(rem_unadjusted, {})),
    'Adjusted': rem_contrast(rem_baseline, rem_slope_vector(rem_baseline, {}))}).T
display(rem_baseline_results.round(4))
rem_lines(rem_baseline, {'Adjusted baseline': {}}, 'Screen time and REM: the baseline association')
baseline_effect = rem_baseline_results.loc['Adjusted'].to_dict()
rem_interpretations['baseline'] = (
    '**Outcome:** Each additional 30 minutes of phone use is associated with '
    + rem_effect_sentence(baseline_effect) + ', holding the listed factors constant. '
    'The unadjusted row shows how the estimate changes after adjustment. '
    'This is the starting screen-time penalty; it assumes the same slope for everyone. '
    'The interaction models below ask whether that slope varies, rather than merely whether '
    'one group has a lower overall REM level. Shading in the plots represents a 95% confidence '
    'interval for the fitted mean, not the spread of individual sleep outcomes.')
```

```python
display(Markdown('**My starting point:** Another 30 phone minutes corresponds to '
    + rem_effect_sentence(baseline_effect)
    + ' after adjustment. I check whether its interval includes zero before calling this a REM-percentage penalty.'))
```

### Brightness and the screen-time slope

This compares the slope after a 25-percentage-point increase in brightness.

```python
rem_brightness_model = smf.ols(rem_main_formula + ' + screen_30:brightness_25', data=rem_analysis).fit(cov_type='HC3')
rem_models['brightness'] = rem_brightness_model
brightness_scenarios = {'25% brightness': {'brightness_25': -1},
                        '50% brightness': {'brightness_25': 0},
                        '75% brightness': {'brightness_25': 1}}
display(rem_slope_table(rem_brightness_model, brightness_scenarios).round(4))
brightness_difference = rem_contrast(rem_brightness_model,
    rem_slope_vector(rem_brightness_model, {'brightness_25': 1}) -
    rem_slope_vector(rem_brightness_model, {'brightness_25': 0}))
display(pd.DataFrame([brightness_difference], index=['Slope change per +25 brightness percentage points']).round(4))
rem_interaction_test(rem_brightness_model, 'brightness')
rem_lines(rem_brightness_model, brightness_scenarios, 'Does brightness change the screen-time slope?')
rem_interpretations['brightness'] = (
    '**Outcome:** Increasing brightness by 25 percentage points changes the REM association '
    'per additional 30 phone minutes by ' + rem_effect_sentence(brightness_difference) + '. '
    'A negative difference means a larger screen-time penalty at higher brightness; a positive '
    'difference means a smaller penalty. The rows show the actual slopes at 25%, 50%, and 75% '
    'brightness. A vertical separation between lines alone is a brightness main effect, '
    'not evidence that brightness amplifies prolonged phone use.')
```

```python
display(Markdown('**What I estimate:** The change in slope is ' + rem_effect_sentence(brightness_difference) + '. If the interval includes zero, I do not have a clear difference in slopes.'))
```

### Blue-light filter and the screen-time slope

This compares the slope with the filter ON versus OFF.

```python
rem_filter_model = smf.ols(rem_main_formula + ' + screen_30:filter_on', data=rem_analysis).fit(cov_type='HC3')
rem_models['filter'] = rem_filter_model
filter_scenarios = {'Filter OFF': {'filter_on': 0}, 'Filter ON': {'filter_on': 1}}
display(rem_slope_table(rem_filter_model, filter_scenarios).round(4))
filter_difference = rem_contrast(rem_filter_model,
    rem_slope_vector(rem_filter_model, {'filter_on': 1}) -
    rem_slope_vector(rem_filter_model, {'filter_on': 0}))
display(pd.DataFrame([filter_difference], index=['Slope change: ON minus OFF']).round(4))
display(rem_analysis.assign(long_phone=rem_analysis.screen_30 > 3).groupby('filter_on').agg(
    n=('rem_pct', 'size'), over_90_minutes=('long_phone', 'sum')))
rem_interaction_test(rem_filter_model, 'filter')
rem_lines(rem_filter_model, filter_scenarios, 'Does filtering change the screen-time slope?')
rem_interpretations['filter'] = (
    '**Outcome:** Filter ON versus OFF changes the REM association per additional 30 phone '
    'minutes by ' + rem_effect_sentence(filter_difference) + '. A positive difference would '
    'suggest attenuation of a negative screen-time slope; a negative difference would suggest '
    'a larger penalty. This tests screen time × filter, not brightness × filter. '
    'Filter status is binary, and the group counts show how much long-use data support each '
    'comparison. Filter users may differ in unmeasured ways, so the estimate is not a causal benefit.')
```

```python
display(Markdown('**What I estimate:** The change in slope is ' + rem_effect_sentence(filter_difference) + '. If the interval includes zero, I do not have a clear difference in slopes.'))
```

### Caffeine and the screen-time slope

```python
rem_caffeine_model = smf.ols(rem_main_formula + ' + screen_30:caffeine_100', data=rem_analysis).fit(cov_type='HC3')
rem_models['caffeine'] = rem_caffeine_model
caffeine_scenarios = {'0 mg after 5 pm': {'caffeine_100': 0},
                      '100 mg after 5 pm': {'caffeine_100': 1},
                      '200 mg after 5 pm': {'caffeine_100': 2}}
display(rem_slope_table(rem_caffeine_model, caffeine_scenarios).round(4))
caffeine_difference = rem_contrast(rem_caffeine_model,
    rem_slope_vector(rem_caffeine_model, {'caffeine_100': 1}) -
    rem_slope_vector(rem_caffeine_model, {'caffeine_100': 0}))
display(pd.DataFrame([caffeine_difference], index=['Slope change per +100 mg evening caffeine']).round(4))
rem_interaction_test(rem_caffeine_model, 'caffeine')
rem_lines(rem_caffeine_model, caffeine_scenarios, 'Does evening caffeine change the screen-time slope?')
rem_interpretations['caffeine'] = (
    '**Outcome:** An additional 100 mg of caffeine after 5 pm changes the REM association '
    'per additional 30 phone minutes by ' + rem_effect_sentence(caffeine_difference) + '. '
    'A negative difference would indicate a larger screen-time penalty with more caffeine. '
    'Caffeine can have a separate association with REM even if this interaction is weak. '
    'The lines assume a linear change in slope with caffeine dose; the 200 mg scenario may '
    'have less support than the lower-dose scenarios, so it should not be treated as an intervention estimate.')
```

### App choice and the screen-time slope

For apps, I test all the app interactions together. Sorting the estimated slopes can show their order in this sample, but it cannot establish that one app is worse than another.

Because I test four interaction questions, I also show **Benjamini–Hochberg (BH) adjusted p-values**. This is a way to limit false discoveries when testing several questions. It does not correct every interval or every comparison elsewhere in the notebook.

```python
rem_app_model = smf.ols(rem_main_formula + ' + screen_30:C(bedtime_app)', data=rem_analysis).fit(cov_type='HC3')
rem_models['app'] = rem_app_model
app_scenarios = {name: {'bedtime_app': name} for name in sorted(rem_analysis.bedtime_app.unique())}
app_slopes = rem_slope_table(rem_app_model, app_scenarios)
app_support = rem_analysis.assign(long_phone=rem_analysis.screen_30 > 3).groupby('bedtime_app').agg(
    n=('rem_pct', 'size'), over_90_minutes=('long_phone', 'sum'),
    phone_minutes_min=('screen_30', lambda s: s.min() * 30),
    phone_minutes_max=('screen_30', lambda s: s.max() * 30))
display(app_slopes.join(app_support).sort_values('REM pp per +30 min').round(4))
app_joint_p = rem_interaction_test(rem_app_model, 'app')
print(f'Joint robust F-test of all screen-time-by-app terms: p = {app_joint_p:.4g}')
rem_lines(rem_app_model, app_scenarios, 'Does the screen-time slope differ by app?')
app_steepest = app_slopes['REM pp per +30 min'].idxmin()
app_flattest = app_slopes['REM pp per +30 min'].idxmax()
rem_interpretations['app'] = (
    '**Outcome:** The most negative estimated slope is for **' + app_steepest + '** ('
    + rem_effect_sentence(app_slopes.loc[app_steepest].to_dict()) + '), and the least negative '
    'is for **' + app_flattest + '** (' + rem_effect_sentence(app_slopes.loc[app_flattest].to_dict())
    + f'). The joint test asks whether any app slope differs (p = {app_joint_p:.4g}). '
    'Ranking the point estimates alone does not establish that the apps differ: the joint '
    'interaction test, intervals, and sample counts matter. These are broad primary-app '
    'categories, not measurements of content, scrolling intensity, or minutes spent in each app.')

# Correct across the four planned interaction hypotheses, including the joint app test.
rem_interaction_summary = pd.DataFrame(rem_interaction_tests).T
rem_interaction_summary['BH-adjusted p-value'] = multipletests(
    rem_interaction_summary['interaction p-value'], method='fdr_bh')[1]
rem_interaction_summary['change in adjusted R2 vs baseline'] = (
    rem_interaction_summary['adjusted R2'] - rem_baseline.rsquared_adj)
display(rem_interaction_summary.round(5))
for name in rem_interaction_tests:
    p = rem_interaction_summary.loc[name, 'interaction p-value']
    q = rem_interaction_summary.loc[name, 'BH-adjusted p-value']
    conclusion = ('There is evidence of a slope difference after accounting for the four planned tests.'
                  if q < 0.05 else 'There is no clear evidence of a slope difference after accounting for the four planned tests; this does not prove the slopes are identical.')
    rem_interpretations[name] += f' **Interaction evidence:** p = {p:.4g}; BH-adjusted p = {q:.4g}. ' + conclusion
supported = rem_interaction_summary.index[rem_interaction_summary['BH-adjusted p-value'] < 0.05].tolist()
rem_interpretations['summary'] = (
    '**Answer to the question:** ' +
    ('The interactions with evidence of different screen-time slopes are: ' + ', '.join(supported) + '. '
     if supported else 'None of these four interaction tests clearly identifies a behavior that makes the screen-time REM penalty larger or smaller in this dataset. ')
    + 'Distinguish a factor associated with lower REM overall from a factor that changes the penalty '
    'for an extra 30 minutes of screen time. The individual interpretations above quantify those '
    'slope changes. These models impose a constant screen-time slope within each exposure profile; '
    'they do not establish a special threshold for long use. Sparse combinations at the exposure '
    'extremes, nonlinear relationships, unmeasured confounding, and unknown dataset provenance '
    'limit the conclusions. The four-interaction correction does not adjust individual slope '
    'confidence intervals or establish pairwise differences between apps. REM percentage alone '
    'also does not capture the full sleep penalty, such as reduced total sleep or REM minutes.')
```

```python
display(Markdown(
    '**What I take from the slope checks:** '
    + ('The factors with BH-adjusted p-values below 0.05 are: ' + ', '.join(supported) + '. '
       if supported else 'None of the four factors has a BH-adjusted interaction p-value below 0.05. ')
    + 'A small positive or negative estimate on its own is not enough for me to say that a habit changes the relationship. '
    + 'No clear difference also does not prove that the slopes are identical. These checks concern REM percentage; '
    + 'they do not rule out differences in total sleep or stage minutes.'))
```

## 3. Stacking: What happens when habits occur together?

So far I have looked at duration and individual conditions. Next I want to see what happens when several potentially unfavorable habits appear in the same record.

I give one flag each for phone use of at least 60 minutes, brightness of at least 75%, any caffeine after 5 pm, a filter that is off, and less than 30 minutes of physical activity. Activity is a daily measure, so this is broader than bedtime behavior alone.

These are choices for exploring the data. Calling something a flag does not establish that it is harmful, and giving each flag one point does not mean all five matter equally. In particular, the filter flag does not assume that filtering improves sleep. Sleep outcomes are kept out of the score so I do not define a group by the same outcome I am trying to compare. App choice stays in the context analysis because I do not have a justified way to label one app as unfavorable.

```python
# Independent of earlier derived frames; run after loading df.
habit_inputs = ['bedtime_phone_minutes', 'screen_brightness_pct',
                'blue_light_filter_active', 'caffeine_post_5pm_mg', 'physical_activity_min']
habit_outcomes = ['total_sleep_hours', 'sleep_latency_min', 'next_day_fatigue_score']
habit_controls = ['age', 'gender', 'occupation_type', 'chronotype', 'primary_bedtime_app']
habits = df.dropna(subset=habit_inputs + habit_outcomes + habit_controls +
                         ['sleep_debt_category', 'user_id']).copy()
assert habits.user_id.is_unique, 'Repeated participants need clustered inference.'
assert habits.blue_light_filter_active.isin([0, 1]).all()
habit_debt_order = ['Optimal Recovery', 'Mild Deficit', 'Moderate Debt', 'Severe Sleep Debt']
assert habits.sleep_debt_category.isin(habit_debt_order).all()
habits['severe_debt'] = habits.sleep_debt_category.eq('Severe Sleep Debt').astype(int)
habit_labels = {'high_screen': 'High screen time', 'high_brightness': 'High brightness',
                'evening_caffeine': 'Evening caffeine', 'filter_off': 'Filter off',
                'low_activity': 'Low activity'}
habit_flags = list(habit_labels)

def make_habit_flags(data, screen=60, brightness=75, caffeine=0, activity=30):
    return pd.DataFrame({
        'high_screen': data.bedtime_phone_minutes.ge(screen),
        'high_brightness': data.screen_brightness_pct.ge(brightness),
        'evening_caffeine': data.caffeine_post_5pm_mg.gt(caffeine),
        'filter_off': data.blue_light_filter_active.eq(0),
        'low_activity': data.physical_activity_min.lt(activity)}, index=data.index).astype(int)

habits[habit_flags] = make_habit_flags(habits)
habits['habit_score'] = habits[habit_flags].sum(axis=1)
habit_definitions = pd.DataFrame({
    'flag': list(habit_labels.values()),
    'one point when': ['Phone use >= 60 min', 'Brightness >= 75%', 'Caffeine after 5 pm > 0 mg',
                      'Blue-light filter = 0 (off)', 'Physical activity < 30 min'],
    'n flagged': habits[habit_flags].sum().values,
    'flagged (%)': habits[habit_flags].mean().values * 100})
print(f'{len(habits):,} rows retained; {len(df)-len(habits):,} excluded for missing analysis fields.')
display(habit_definitions.round(1))

def habit_summary(data, by):
    result = data.groupby(by, observed=True).agg(
        n=('user_id', 'size'), sleep_hours=('total_sleep_hours', 'mean'),
        sleep_median=('total_sleep_hours', 'median'), sleep_sd=('total_sleep_hours', 'std'),
        latency_min=('sleep_latency_min', 'mean'), fatigue=('next_day_fatigue_score', 'mean'),
        severe_debt_pct=('severe_debt', lambda x: x.mean() * 100))
    # Approximate intervals for the mean, not individual prediction intervals.
    se = result.sleep_sd / np.sqrt(result.n)
    result['sleep_ci_low'] = result.sleep_hours - 1.96 * se
    result['sleep_ci_high'] = result.sleep_hours + 1.96 * se
    result['small_group'] = result.n < 30
    return result
```

### One flag at a time

I first compare people with each flag to people without it. A negative sleep difference means the flagged group slept less on average. The same person can belong to several flagged groups, so I cannot add these differences together.

```python
habit_individual = pd.concat({habit_labels[f]: habit_summary(habits, f)
                              for f in habit_flags}, names=['habit', 'flag'])
display(habit_individual.round(2))
habit_differences = pd.DataFrame({label:
    habit_individual.loc[(label, 1), ['sleep_hours', 'latency_min', 'fatigue', 'severe_debt_pct']] -
    habit_individual.loc[(label, 0), ['sleep_hours', 'latency_min', 'fatigue', 'severe_debt_pct']]
    for label in habit_labels.values()}).T.astype(float)
habit_differences.columns = ['Sleep difference (h)', 'Latency difference (min)',
                             'Fatigue difference', 'Severe-debt difference (pp)']
display(habit_differences.round(2))
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for ax, col in zip(axes, ['Sleep difference (h)', 'Severe-debt difference (pp)']):
    habit_differences[col].plot.barh(ax=ax, color='#527fa3')
    ax.axvline(0, color='black', linewidth=1)
    ax.set(xlabel=col, ylabel='')
fig.suptitle('Individual habits: unadjusted flagged-minus-unflagged differences')
plt.tight_layout(); plt.show()
```

### Counting flags

I compare both the exact count from zero to five and three broader bands: 0–1, 2–3, and 4–5. I check the number of people in each group alongside sleep, fatigue, and debt. The error bars are approximate 95% intervals for the group mean, not a prediction for an individual.

```python
habits['score_band'] = pd.cut(habits.habit_score, [-1, 1, 3, 5], labels=['0–1 flags', '2–3 flags', '4–5 flags'])
habit_count_table = habit_summary(habits, 'habit_score')
habit_band_table = habit_summary(habits, 'score_band')
display(habit_count_table.round(2), habit_band_table.round(2))
habit_debt_mix = pd.crosstab(habits.habit_score, habits.sleep_debt_category,
    normalize='index').reindex(columns=habit_debt_order, fill_value=0) * 100
fig, axes = plt.subplots(1, 3, figsize=(17, 4))
axes[0].errorbar(habit_count_table.index, habit_count_table.sleep_hours,
    yerr=1.96 * habit_count_table.sleep_sd / np.sqrt(habit_count_table.n), fmt='o-', capsize=4)
axes[0].set(xlabel='Number of flags', ylabel='Mean sleep (hours)', title='Approximate 95% mean intervals')
sns.boxplot(data=habits, x='score_band', y='next_day_fatigue_score', color='#9fc4d5', showfliers=False, ax=axes[1])
axes[1].set(xlabel='', ylabel='Fatigue score', title='Fatigue distribution (outliers hidden)')
habit_debt_mix.plot.bar(stacked=True, colormap='YlOrRd', ax=axes[2])
axes[2].set(xlabel='Number of flags', ylabel='Observations (%)', title='Sleep-debt category mix')
axes[2].legend(fontsize=7)
plt.tight_layout(); plt.show()
```

### Separating screen habits from other habits

I also split the count into screen habits (duration, brightness, filter) and other habits (caffeine, activity). The heatmaps show how these combinations line up with sleep and debt. Each box’s group size helps me judge how much data are behind the average. The screen side has three possible flags and the other side has two, so their totals do not have equal weight.

```python
habits['screen_score'] = habits[['high_screen', 'high_brightness', 'filter_off']].sum(axis=1)
habits['other_score'] = habits[['evening_caffeine', 'low_activity']].sum(axis=1)
habit_domains = habit_summary(habits, ['screen_score', 'other_score'])
display(habit_domains.round(2))
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, value, title in zip(axes, ['n', 'sleep_hours', 'severe_debt_pct'],
                          ['Group size', 'Mean sleep (hours)', 'Severe debt (%)']):
    sns.heatmap(habit_domains[value].unstack(), annot=True, fmt='.0f' if value=='n' else '.2f',
                cmap='Blues', ax=ax)
    ax.set(title=title, xlabel='Other-behavior flags (0–2)', ylabel='Screen flags (0–3)')
plt.tight_layout(); plt.show()
```

### Same count, different combinations

Two flags can describe very different routines. I list the exact combinations and then compare profiles with exactly two flags. I show the plot only for groups with at least 30 observations. That is a display rule, not a guarantee that the estimates are precise. I do not treat the ordering of the averages as proof that every profile differs from every other one.

```python
habits['profile'] = habits[habit_flags].apply(
    lambda row: ' + '.join(habit_labels[f] for f in habit_flags if row[f]) or 'No flags', axis=1)
habit_profiles = habit_summary(habits, ['habit_score', 'profile']).sort_index()
with pd.option_context('display.max_rows', 40, 'display.max_colwidth', 100):
    display(habit_profiles.round(2))
habit_two = habit_profiles.xs(2, level='habit_score').query('n >= 30').sort_values('sleep_hours')
fig, ax = plt.subplots(figsize=(11, 6))
ax.errorbar(habit_two.sleep_hours, np.arange(len(habit_two)),
    xerr=1.96 * habit_two.sleep_sd / np.sqrt(habit_two.n), fmt='o', capsize=3)
ax.set_yticks(np.arange(len(habit_two)), [f'{name} (n={int(n)})' for name,n in zip(habit_two.index,habit_two.n)])
ax.set(xlabel='Mean sleep (hours), approximate 95% CI', title='Exactly two flags: composition matters')
plt.tight_layout(); plt.show()
```

### Checking my cutoff choices

I repeat the band comparison using stricter screen cutoffs and cutoffs based on the sample’s upper or lower quarter. Caffeine stays “any after 5 pm,” and filter status stays unchanged. If the pattern changes a lot, I should be cautious about a conclusion that depends on my first choices. A score of three under one definition is not necessarily the same set of behaviors under another.

```python
habit_thresholds = {
    'Main: 60 min / 75% / <30 active min': (60, 75, 30),
    'Stricter screens: 90 min / 80% / <30 active min': (90, 80, 30),
    'Sample quartiles: upper screens / lower activity': (
        habits.bedtime_phone_minutes.quantile(.75), habits.screen_brightness_pct.quantile(.75),
        habits.physical_activity_min.quantile(.25))}
habit_sensitivity_parts = {}
for name, (screen, brightness, activity) in habit_thresholds.items():
    variant = habits.copy()
    variant['variant_score'] = make_habit_flags(variant, screen=screen, brightness=brightness,
                                               activity=activity).sum(axis=1)
    variant['variant_band'] = pd.cut(variant.variant_score, [-1, 1, 3, 5], labels=['0–1 flags', '2–3 flags', '4–5 flags'])
    habit_sensitivity_parts[name] = habit_summary(variant, 'variant_band')
display(pd.DataFrame(habit_thresholds, index=['Screen >= min', 'Brightness >= %', 'Activity < min']).T)
habit_sensitivity = pd.concat(habit_sensitivity_parts, names=['definition', 'band'])
display(habit_sensitivity.round(2))
```

### Do pairs show an extra combined difference?

A worse average at higher flag counts does not automatically mean that habits amplify one another. I first include all five flags in one model, along with age, gender, occupation, chronotype, and app.

Then I add three pairs: high screen time × high brightness, high screen time × evening caffeine, and high brightness × filter off. The interaction estimate is the extra sleep difference beyond adding the two individual terms. I apply the BH adjustment to these three tests and show how many observations support each pair.

```python
habit_adjusters = 'age + C(gender) + C(occupation_type) + C(chronotype) + C(primary_bedtime_app)'
habit_formula = 'total_sleep_hours ~ ' + ' + '.join(habit_flags) + ' + ' + habit_adjusters
habit_additive = smf.ols(habit_formula, data=habits).fit(cov_type='HC3')
habit_adjusted = pd.DataFrame({
    'Adjusted sleep difference (h)': habit_additive.params[habit_flags],
    '95% CI low': habit_additive.conf_int().loc[habit_flags, 0],
    '95% CI high': habit_additive.conf_int().loc[habit_flags, 1]})
display(habit_adjusted.rename(index=habit_labels).round(3))
habit_pairs = ['high_screen:high_brightness', 'high_screen:evening_caffeine', 'high_brightness:filter_off']
habit_interaction = smf.ols(habit_formula + ' + ' + ' + '.join(habit_pairs), data=habits).fit(cov_type='HC3')
habit_interaction_table = pd.DataFrame({
    'Extra joint sleep difference (h)': habit_interaction.params[habit_pairs],
    '95% CI low': habit_interaction.conf_int().loc[habit_pairs, 0],
    '95% CI high': habit_interaction.conf_int().loc[habit_pairs, 1],
    'p': habit_interaction.pvalues[habit_pairs]})
habit_interaction_table['BH-adjusted p'] = multipletests(habit_interaction_table.p, method='fdr_bh')[1]
display(habit_interaction_table.round(4))
# Show support for each 2x2 comparison rather than hiding sparse cells.
for pair in habit_pairs:
    first, second = pair.split(':')
    display(pd.crosstab(habits[first], habits[second]).rename_axis(index=first, columns=second))
```

```python
habit_low = habit_band_table.loc['0–1 flags']
habit_high = habit_band_table.loc['4–5 flags']
habit_supported = habit_interaction_table.index[habit_interaction_table['BH-adjusted p'] < .05].tolist()
habit_sensitivity_deltas = {
    name: table.loc['4–5 flags', 'sleep_hours'] - table.loc['0–1 flags', 'sleep_hours']
    for name, table in habit_sensitivity_parts.items()}
display(Markdown(
    f"**What I notice about stacking:** The 0–1-flag group has {int(habit_low.n):,} observations and averages "
    f"{habit_low.sleep_hours:.2f} sleep hours. The 4–5-flag group has {int(habit_high.n):,} observations and averages "
    f"{habit_high.sleep_hours:.2f} hours, a difference of {habit_high.sleep_hours-habit_low.sleep_hours:+.2f} hours. "
    f"Among the two-flag profiles shown, average sleep ranges from {habit_two.sleep_hours.min():.2f} "
    f"to {habit_two.sleep_hours.max():.2f} hours. The particular combination matters when describing the groups.\n\n"
    + 'Across my cutoff choices, the 4–5 minus 0–1 differences are: '
    + '; '.join(f'{name}: {value:+.2f} hours' for name, value in habit_sensitivity_deltas.items()) + '.\n\n'
    + ('The pairs with BH-adjusted p < 0.05 are ' + ', '.join(habit_supported) + '. '
       if habit_supported else 'None of the three pairs has BH-adjusted p < 0.05. ')
    + 'I keep the count pattern separate from evidence of an extra combined effect.'))
```

## 4. Sleep: How do sleep stages and debt differ across profiles?

The stacking section mainly compared total sleep. Now I connect those same flags and profiles to REM, deep sleep, and debt. This is the added link between the original notebook’s habit analysis and its sleep-stage analysis.

I reuse the existing flag assignments and remove records missing either stage percentage for this comparison. I show the number retained, since different missing-data requirements can change the sample. I compare both percentages and calculated minutes: less total sleep can mean fewer stage minutes even if the stage’s percentage changes little.

These are unadjusted group summaries. They cannot separate the effects of each habit or differences in the people who have them.

```python
profile_sleep = habits.dropna(subset=['rem_sleep_pct', 'deep_sleep_pct']).copy()
profile_sleep['rem_minutes'] = profile_sleep.total_sleep_hours * 60 * profile_sleep.rem_sleep_pct / 100
profile_sleep['deep_minutes'] = profile_sleep.total_sleep_hours * 60 * profile_sleep.deep_sleep_pct / 100
print(f'{len(profile_sleep):,} records retained for stage comparisons; {len(habits)-len(profile_sleep):,} excluded.')
profile_measures = ['total_sleep_hours', 'rem_sleep_pct', 'deep_sleep_pct', 'rem_minutes', 'deep_minutes']

def summarize_profile_sleep(by):
    grouped = profile_sleep.groupby(by, observed=True)
    summary = grouped[profile_measures].mean()
    summary.insert(0, 'n', grouped.size())
    summary['severe_debt_pct'] = grouped.severe_debt.mean() * 100
    return summary

stage_by_count = summarize_profile_sleep('habit_score')
stage_by_band = summarize_profile_sleep('score_band')
stage_by_profile = summarize_profile_sleep(['habit_score', 'profile'])
display(stage_by_count.round(2), stage_by_band.round(2))
with pd.option_context('display.max_rows', 40, 'display.max_colwidth', 90):
    display(stage_by_profile.round(2))

fig, axes = plt.subplots(2, 2, figsize=(13, 9))
for ax, measure in zip(axes.flat, ['rem_sleep_pct', 'deep_sleep_pct', 'rem_minutes', 'deep_minutes']):
    sns.boxplot(data=profile_sleep, x='score_band', y=measure, color='#9fc4d5', showfliers=False, ax=ax)
    ax.set(xlabel='Behavioral flag band', ylabel=story_outcomes[measure])
fig.suptitle('Sleep stages across behavioral flag bands')
plt.tight_layout()
plt.show()

profile_debt_mix = pd.crosstab(profile_sleep.score_band, profile_sleep.sleep_debt_category,
    normalize='index').reindex(columns=story_debt_order, fill_value=0) * 100
profile_debt_mix.plot.bar(stacked=True, colormap='YlOrRd', figsize=(10, 5))
plt.ylabel('Participants (%)')
plt.xlabel('Behavioral flag band')
plt.title('Recorded sleep debt across the same flag bands')
plt.xticks(rotation=0)
plt.legend(title='Debt category', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### Accounting for other recorded differences

I fit a separate fatigue model for each sleep measure, first by itself and then with the original phone and background variables. The units differ: one hour for total sleep, 30 minutes for stage duration, and one percentage point for stage percentages.

I also fit one joint model with total sleep and both stage percentages. I do not put calculated stage minutes into that model because they already combine total sleep and percentages. At fixed total sleep, increasing one stage’s percentage leaves less time for other stages.

Debt gets its own model, using Optimal Recovery as the reference. I leave total sleep out of that debt comparison because the debt categories may already contain overlapping sleep information. I do not know their creation rules. These are exploratory linear models for a bounded fatigue score, not tested tools for predicting an individual’s fatigue.

```python
story_fatigue_models, story_fatigue_rows = {}, []
story_sleep_steps = {'total_sleep_hours': 1, 'rem_sleep_pct': 1, 'deep_sleep_pct': 1,
                     'rem_minutes': 30, 'deep_minutes': 30}
for outcome, step in story_sleep_steps.items():
    for adjustment, rhs in [('Unadjusted', outcome), ('Adjusted', outcome + ' + ' + story_rhs)]:
        model = smf.ols('next_day_fatigue_score ~ ' + rhs, data=phone_story).fit(cov_type='HC3')
        story_fatigue_models[(outcome, adjustment)] = model
        low, high = model.conf_int().loc[outcome] * step
        story_fatigue_rows.append({'sleep measure': story_outcomes[outcome], 'increment': step,
            'model': adjustment, 'fatigue difference': model.params[outcome] * step,
            '95% CI lower': low, '95% CI upper': high})
story_fatigue_effects = pd.DataFrame(story_fatigue_rows)
display(story_fatigue_effects.set_index(['sleep measure', 'model']).round(3))

story_joint_sleep_model = smf.ols('next_day_fatigue_score ~ total_sleep_hours + rem_sleep_pct + '
    'deep_sleep_pct + ' + story_rhs, data=phone_story).fit(cov_type='HC3')
story_joint_terms = ['total_sleep_hours', 'rem_sleep_pct', 'deep_sleep_pct']
story_joint_results = pd.DataFrame({'fatigue difference': story_joint_sleep_model.params[story_joint_terms],
    '95% CI lower': story_joint_sleep_model.conf_int().loc[story_joint_terms, 0],
    '95% CI upper': story_joint_sleep_model.conf_int().loc[story_joint_terms, 1]})
display(story_joint_results.round(3))

story_debt_fatigue_model = smf.ols('next_day_fatigue_score ~ '
    'C(sleep_debt_category, Treatment(reference="Optimal Recovery")) + ' + story_rhs,
    data=phone_story).fit(cov_type='HC3')
story_debt_terms = [term for term in story_debt_fatigue_model.params.index if term.startswith('C(sleep_debt_category')]
story_debt_fatigue_results = pd.DataFrame({
    'fatigue difference vs Optimal Recovery': story_debt_fatigue_model.params[story_debt_terms],
    '95% CI lower': story_debt_fatigue_model.conf_int().loc[story_debt_terms, 0],
    '95% CI upper': story_debt_fatigue_model.conf_int().loc[story_debt_terms, 1]})
story_debt_fatigue_results.index = [term.split('[T.')[1].rstrip(']') for term in story_debt_terms]
display(story_debt_fatigue_results.round(3))

fatigue_sentences = []
for outcome in ['total_sleep_hours', 'rem_minutes', 'deep_minutes']:
    row = story_fatigue_effects.loc[(story_fatigue_effects['sleep measure'] == story_outcomes[outcome]) &
                                   (story_fatigue_effects['model'] == 'Adjusted')].iloc[0]
    unit = 'one additional hour of total sleep' if outcome == 'total_sleep_hours' else '30 additional minutes of ' + ('REM' if outcome == 'rem_minutes' else 'deep sleep')
    fatigue_sentences.append(f"{unit}: **{row['fatigue difference']:+.2f} fatigue points** "
        f"(95% CI {row['95% CI lower']:+.2f} to {row['95% CI upper']:+.2f})")
story_notes['part3'] = ('**What this shows:** In separate adjusted models, ' + '; '.join(fatigue_sentences) + '. '
    'These stage-duration associations are not independent of total sleep: inspect the joint '
    'model to see whether stage percentages add an association at a fixed total duration. '
    'At fixed total sleep, a percentage-point increase in one stage also implies less time '
    'in other stages, so its coefficient is a composition contrast. The debt table compares '
    'recorded categories with Optimal Recovery, adjusting for phone behaviors and background '
    'variables but not total sleep. No category-generation rule is known, so the debt association '
    'may partly repeat information already present in sleep duration. These are exploratory '
    'linear associations with a bounded fatigue score, not validated predictions or causal effects.')
story_notes['ending'] = (
    '**Phone behavior → Sleep outcomes → Next-day fatigue:** The strongest narrative is '
    'about how much sleep people get, while distinguishing stage duration from stage percentage. '
    'The first part quantifies the phone–sleep association; the second compares conditions '
    'at the same one-hour exposure; the third relates sleep differences to fatigue. '
    'Together they describe a connected pattern, but do not show that sleep mediates a causal '
    'phone effect. The models use the same observational sample and cannot establish temporal '
    'mechanisms, eliminate unmeasured confounding, or establish that changing a phone setting '
    'will improve sleep. Small point estimates with intervals spanning zero should not be '
    'presented as proven improvements or harms.')
```

```python
display(Markdown('**What I find in the separate models:** ' + '; '.join(fatigue_sentences)
    + '. These are adjusted associations, and the stage-minute results still overlap with total sleep.'))
for term, label in [('total_sleep_hours', 'one more hour of total sleep'),
                    ('rem_sleep_pct', 'one more REM percentage point'),
                    ('deep_sleep_pct', 'one more deep-sleep percentage point')]:
    row = story_joint_results.loc[term]
    display(Markdown(
        f"**In the joint model, {label}:** {row['fatigue difference']:+.3f} fatigue points "
        f"(95% CI {row['95% CI lower']:+.3f} to {row['95% CI upper']:+.3f}). "
        + ('The interval includes zero, so I do not see a clear separate association here.'
           if row['95% CI lower'] <= 0 <= row['95% CI upper']
           else 'The interval does not include zero, although this still does not establish cause and effect.')))
for category, row in story_debt_fatigue_results.iterrows():
    display(Markdown(
        f"**{category} compared with Optimal Recovery:** "
        f"{row.iloc[0]:+.2f} adjusted fatigue points (95% CI {row.iloc[1]:+.2f} to {row.iloc[2]:+.2f})."))
```

My main lesson is that I need to be clear about what I am measuring. Total sleep, the percentage spent in a stage, the minutes spent in that stage, and recorded debt categories answer different questions. I also need to distinguish a difference between groups from evidence that two habits amplify each other.

The summary mixes adjusted model estimates with unadjusted group averages; its comparison column identifies which is which. The context checks concern REM percentage, while the same-duration comparisons earlier cover more sleep outcomes. I should not turn a weak slope result into a claim that a habit has no relationship with sleep at all.

These observations do not establish that phone use caused the sleep differences, or that sleep caused the fatigue differences. Unmeasured factors, the unknown data-generation process, sparse combinations, and model assumptions could affect the results. Checking several outcomes also gives me more chances to find an apparent pattern by accident.

As a next step, I would want better information about where the data came from, repeated observations of the same people, and actual phone-use and bedtime timestamps. I would also check whether curved relationships fit better than straight lines before trying to identify a threshold.
